```python
product = "PACE_OCI_L3M_SFREFL"
date_ini, date_end= '2025-05-02', '2025-05-31'
# Level 3 data
results = earthaccess.search_data(
    short_name = product,
    temporal = (date_ini, date_end),
    granule_name="*R32*0p1deg*"
)
len(results)
```

```python
fileset = earthaccess.open(results)
```

```python
import xarray as xr
ds = xr.open_dataset(fileset[0], chunks={})
ds
```

```python
print(ds.attrs)
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

banda_nir = ds.rhos.sel(wavelength=865, method='nearest')

plt.figure(figsize=(12, 6))
ax = plt.axes(projection=ccrs.PlateCarree())
ax.coastlines()


banda_nir.plot(ax=ax, x='lon', y='lat', cmap='gray', robust=True)
plt.title("Reflectancia en 865 nm (Infrarrojo Cercano)")
plt.show()
```
