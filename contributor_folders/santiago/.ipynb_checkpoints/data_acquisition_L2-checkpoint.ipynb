```python
import xarray as xr
import numpy as np

def extract_transect(ds, var_name, lats, lons):
    """
    Extracts a transect from a Dataset using the variable name and 
    geographic coordinates.
    
    Parameters:
    -----------
    ds : xarray.Dataset
        The source dataset containing 'latitude', 'longitude', and the variable.
    var_name : str
        The name of the variable to extract (e.g., 'rhos', 'Rrs', 'chlor_a').
    lats : list or np.array
        Target latitudes for the transect.
    lons : list or np.array
        Target longitudes for the transect.
        
    Returns:
    --------
    selection : xarray.DataArray
        The extracted variable with 'points' dimension and lat/lon coords.
    """
    # 1. Pull coordinate grids and the target variable
    lat_vals = ds["latitude"].values
    lon_vals = ds["longitude"].values
    data_da = ds[var_name]

    # 2. Compute nearest pixels (Manhattan distance)
    rows, cols = [], []
    for lat, lon in zip(lats, lons):
        dist = np.abs(lat_vals - lat) + np.abs(lon_vals - lon)
        i, j = np.unravel_index(dist.argmin(), lat_vals.shape)
        rows.append(i)
        cols.append(j)

    # 3. Extract points using advanced indexing
    # We create DataArrays for the indices to define the new 'points' dimension
    selection = data_da.isel(
        number_of_lines=xr.DataArray(rows, dims="points"), 
        pixels_per_line=xr.DataArray(cols, dims="points")
    ).load()
    
    # 4. Attach the actual lat/lon found as coordinates
    selection = selection.assign_coords(
        longitude=("points", [lon_vals[r, c] for r, c in zip(rows, cols)]),
        latitude=("points", [lat_vals[r, c] for r, c in zip(rows, cols)])
    )

    return selection

# rhos_sel = extract_transect_by_name(ds, "rhos", target_lats, target_lons)
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as colors
import numpy as np

for file_ in fileset[0:]:
    datatree = xr.open_datatree(file_, decode_timedelta=False, chunks={}) 
    
    ds = xr.merge(datatree.to_dict().values())
    ds = ds.set_coords(("longitude", "latitude"))
    ds = ds.unify_chunks()
    # User Input: Define Transect
    lat_start, lon_start = 1.9, 102
    lat_end, lon_end = 2.5, 101.4
    # Define transect
    npts = 20  # increase/decrease for more/less detail along transect
    lats = np.linspace(lat_start, lat_end, npts)
    lons = np.linspace(lon_start, lon_end, npts)
    rhos_sel = extract_transect(ds, 'rhos',lats, lons)
    
    raw_date = file_.full_name.split('.')[1].split('T')[0]
    hour_    = file_.full_name.split('.')[1].split('T')[1]
    formatted_date = f"{raw_date[:4]}-{raw_date[4:6]}-{raw_date[6:]} {hour_[0:2]}:{hour_[2:4]}"
    # 1. Extract coordinates
    # Ensure we are using the values from your rhos_sel DataArray
    lons = rhos_sel.longitude.values
    lats = rhos_sel.latitude.values
    
    # 2. Calculate Euclidean distance in degrees from the first point (index 0)
    # This measures the angular separation on the lat/lon plane
    dist_degrees = np.sqrt((lons - lons[0])**2 + (lats - lats[0])**2)
    
    # 3. Setup the colormap based on distance
    norm = colors.Normalize(vmin=dist_degrees.min(), vmax=dist_degrees.max())
    cmap = cm.get_cmap('plasma') 
    
    fig, ax = plt.subplots(figsize=(12, 7))
    
    # 4. Iterate and plot
    for i, p in enumerate(rhos_sel.points):
        # Select point data (Dask will pull only this vector)
        point_data = rhos_sel.sel(points=p)
        
        # Assign color based on its distance in degrees
        color = cmap(norm(dist_degrees[i]))
        
        # Plotting with English label compatibility
        point_data.plot.line(
            ax=ax, 
            x='wavelength_3d', 
            marker='.', 
            color=color, 
            alpha=0.6, 
            add_legend=False
        )
    
    # 5. Add Colorbar
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = fig.colorbar(sm, ax=ax)
    cbar.set_label('Distance (Degrees °)', rotation=270, labelpad=15)
    
    # English Titles and Labels
    ax.set_title(f"Spectral Signatures: Variation Along the Transect\nDate: {formatted_date}", fontsize=14)
    ax.set_xlabel("Wavelength (nm)")
    ax.set_ylabel("Surface Reflectance (rhos)")
    ax.grid(True, linestyle='--', alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import numpy as np
import matplotlib.cm as cm
import matplotlib.colors as colors

# 1. Extract real coordinates from your DataArray
lons = rhos_sel.longitude.values
lats = rhos_sel.latitude.values

# 2. Calculate distance in degrees (consistent with the spectral plot logic)
dist_degrees = np.sqrt((lons - lons[0])**2 + (lats - lats[0])**2)

# 3. Setup map projection
fig = plt.figure(figsize=(12, 9))
ax = plt.axes(projection=ccrs.PlateCarree())

# High-resolution geographic features
ax.add_feature(cfeature.LAND, facecolor='#f5f5f5', edgecolor='dimgray')
ax.add_feature(cfeature.OCEAN, facecolor='#e3f2fd')
ax.add_feature(cfeature.COASTLINE, linewidth=1)
ax.add_feature(cfeature.BORDERS, linestyle=':', alpha=0.6)
ax.add_feature(cfeature.RIVERS, linewidth=0.5)

# 4. Draw the actual transect
# Base line connecting the points
ax.plot(lons, lats, color='black', linestyle='-', linewidth=1, 
        alpha=0.3, transform=ccrs.PlateCarree(), zorder=1)

# Points colored by distance in degrees (using 'plasma' to match)
norm = colors.Normalize(vmin=dist_degrees.min(), vmax=dist_degrees.max())
path = ax.scatter(lons, lats, c=dist_degrees, cmap='plasma', s=60, 
                  edgecolors='white', linewidth=0.5,
                  transform=ccrs.PlateCarree(), zorder=2)

# 5. Mark Start and End points
ax.plot(lons[0], lats[0], marker='D', color='green', markersize=8, 
        transform=ccrs.PlateCarree(), label='Transect Start', zorder=3)
ax.plot(lons[-1], lats[-1], marker='X', color='red', markersize=10, 
        transform=ccrs.PlateCarree(), label='Transect End', zorder=3)

# 6. Gridlines and labels
gl = ax.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.5, linestyle='--')
gl.top_labels = False
gl.right_labels = False

# 7. Colorbar (identical to the spectral time series)
cbar = plt.colorbar(path, ax=ax, orientation='vertical', shrink=0.6, pad=0.08)
cbar.set_label('Distance (Degrees °)', rotation=270, labelpad=15)

# Set map extent (Zoom into the points with a 0.5 degree margin)
margin = 0.5
ax.set_extent([lons.min() - margin, lons.max() + margin, 
               lats.min() - margin, lats.max() + margin])

plt.title("Geographical Location of Transect Points", fontsize=14)
plt.legend(loc='lower left')

plt.show()
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import numpy as np
lat_start, lon_start = 1.9, 102
lat_end, lon_end = 2.5, 101.4
# 2. Setup the map projection
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

# 3. Add geographic layers for context (High resolution)
ax.add_feature(cfeature.LAND, facecolor='#e0e0e0', edgecolor='gray')
ax.add_feature(cfeature.OCEAN, facecolor='#f0f8ff')
ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
ax.add_feature(cfeature.BORDERS, linestyle=':', alpha=0.5)

# 4. Draw the transect path
# We create intermediate points to show the direction with a color gradient
n_points = 100
lons = np.linspace(lon_start, lon_end, n_points)
lats = np.linspace(lat_start, lat_end, n_points)

# Plotting the line
ax.plot(lons, lats, color='black', linestyle='--', linewidth=1, 
        transform=ccrs.PlateCarree(), zorder=1)

# Scatter plot with colormap to show direction (Start to End)
path = ax.scatter(lons, lats, c=np.arange(n_points), cmap='jet', 
                   s=15, transform=ccrs.PlateCarree(), zorder=2)

# 5. Mark and label the Start and End points
ax.plot(lon_start, lat_start, marker='D', color='green', markersize=8, 
        transform=ccrs.PlateCarree(), label='Start Point', zorder=3)
ax.plot(lon_end, lat_end, marker='X', color='red', markersize=9, 
        transform=ccrs.PlateCarree(), label='End Point', zorder=3)

# 6. Final Layout & Labels
plt.title("PACE/OCI Transect Geographical Scheme", fontsize=14, pad=20)
gl = ax.gridlines(draw_labels=True, dms=True, x_inline=False, y_inline=False,
                  linewidth=0.5, color='gray', alpha=0.3)
gl.top_labels = False
gl.right_labels = False

# Add colorbar to explain the direction
cbar = plt.colorbar(path, ax=ax, orientation='vertical', shrink=0.5, pad=0.05)
cbar.set_label('Sampling Direction (Progression)')
cbar.set_ticks([]) # Remove numbers for a cleaner scheme

# Set map extent (Zoom into the transect area with some padding)
ax.set_extent([np.min([lon_start,lon_end])-.25, 
               np.max([lon_start,lon_end])+.25,
               np.min([lat_start,lat_end])-.25, 
               np.max([lat_start,lat_end])+.25])

plt.legend(loc='lower left')
plt.show()
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import earthaccess
import numpy as np

# --- 1. Configuración de parámetros ---
lon_min, lon_max = -98, -78
lat_min, lat_max = 18, 30.5
date_ini, date_end= '2025-05-02', '2025-05-13'
# Level 2 data
results = earthaccess.search_data(
    short_name = 'PACE_OCI_L2_AOP',
    temporal = (date_ini, date_end),
    bounding_box = (lon_min, lat_min, lon_max, lat_max)
)
len(results)
fileset = earthaccess.open(results);

print(f"Archivos encontrados: {len(results)}")

# --- 3. Procesamiento y Visualización ---
# Usamos un subconjunto para pruebas (fileset[::5])
for file_ in results[::5]:
    try:
        print(f"\nProcesando: {file_.full_name}")
        
        # Extracción de fecha para el título
        # Estructura esperada: ...PACE_OCI.20250513T172205...
        raw_date = file_.full_name.split('.')[1].split('T')[0]
        formatted_date = f"{raw_date[:4]}-{raw_date[4:6]}-{raw_date[6:]}"

        # Abrir datatree (necesario para grupos en PACE L2)
        # Usamos fsspec para abrir directamente desde S3 si es posible
        with earthaccess.open([file_])[0] as f:
            dt = xr.open_datatree(f, decode_timedelta=False, chunks={})
            
            # Combinar grupos: navigation_data (lat/lon) y geophysical_data (avw, etc)
            ds = xr.merge(dt.to_dict().values())
            
            # Establecer coordenadas explícitas para datos de barrido (Swath)
            # Esto resuelve el error de "Dimensions not exist"
            ds = ds.set_coords(("longitude", "latitude"))
            
            # Seleccionar variable
            avw_da = ds["avw"]

            # --- Filtro de área (Transecto) ---
            # Verificamos si hay píxeles con coordenadas dentro de nuestro cuadro
            mask = (
                (ds.latitude >= lat_min) & (ds.latitude <= lat_max) & 
                (ds.longitude >= lon_min) & (ds.longitude <= lon_max)
            )

            if not mask.any():
                print(f"--> Saltando: No hay datos dentro de la BBOX para esta fecha.")
                continue

            # Recortar datos para acelerar el renderizado y evitar outliers globales
            subset = avw_da.where(mask, drop=True)
            
            # Calcular límites de color dinámicos
            vmin = float(subset.quantile(0.01))
            vmax = float(subset.quantile(0.99))

            # --- Configuración del Mapa ---
            fig = plt.figure(figsize=(12, 7))
            ax = plt.axes(projection=ccrs.PlateCarree())
            
            ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
            ax.coastlines(resolution='10m', color='black', linewidth=1)
            gl = ax.gridlines(draw_labels=True, dms=True, x_inline=False, y_inline=False)
            gl.top_labels = False
            gl.right_labels = False

            # Graficar usando las coordenadas de longitud y latitud del sensor
            img = subset.plot(
                x="longitude", 
                y="latitude", 
                ax=ax, 
                cmap="jet", 
                vmin=vmin, 
                vmax=vmax,
                cbar_kwargs={"pad": 0.05, "label": "AVW (nm)"},
                transform=ccrs.PlateCarree()
            )

            ax.set_title(f"PACE OCI - Apparent Visible Wavelength (AVW)\nFecha: {formatted_date}", 
                         fontsize=14, pad=20)
            
            plt.show()
            plt.close()

    except Exception as e:
        print(f"Error al procesar")
```
